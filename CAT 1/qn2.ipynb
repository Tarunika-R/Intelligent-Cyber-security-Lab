### Defining the access control rules - list of dictionary

```python
access_rules = [
    {"rule_no": 1, "role": "admin", "resource": "*", "access_type": "*", "action": "ALLOW"},
    {"rule_no": 2, "role": "admin", "resource": "/exam/grades", "access_type": "READ", "action": "DENY"},
    {"rule_no": 3, "role": "faculty", "resource": "/exam/grades", "access_type": "READ", "action": "ALLOW"},
    {"rule_no": 4, "role": "student", "resource": "/course/materials", "access_type": "READ", "action": "ALLOW"},
    {"rule_no": 5, "role": "*", "resource": "*", "access_type": "*", "action": "DENY"}
]
```

### Building simulator function

```python
def check_access(role, resource, access_type):
    """
    Simulates RBAC access control with first-match-wins policy.
    """
    for rule in access_rules:
        role_match = (rule["role"] == role or rule["role"] == "*")
        resource_match = (rule["resource"] == resource or rule["resource"] == "*")
        access_match = (rule["access_type"] == access_type or rule["access_type"] == "*")

        if role_match and resource_match and access_match:
            # First match found
            return rule  # Return matched rule

    return None  # No rule found (should not happen with default deny)
```

### Running simulation with 5 test samples

```python
test_cases = [
    ("student", "/exam/grades", "READ"),
    ("faculty", "/exam/grades", "READ"),
    ("guest", "/dashboard", "READ"),
    ("admin", "/exam/grades", "READ"),
    ("student", "/course/materials", "READ"),
]
```

### Print result
- access request
- matched rule
- final decision

```python
for role, resource, action in test_cases:
    matched_rule = check_access(role, resource, action)
    final_decision = matched_rule["action"] if matched_rule else "DENY"

    print(f"Access Request: Role='{role}', Resource='{resource}', Action='{action}'")
    if matched_rule:
        print(f"  Matched Rule: #{matched_rule['rule_no']} -> {matched_rule}")
    else:
        print("  Matched Rule: None")
    print(f"  Final Decision: {final_decision}")
    print("-" * 50)
```
